```python
import numpy as np

support = np.array([1, 2, 3, 4, 5])
c = 1.0 / np.sum(support)
pmf = c * support

cdf_3 = np.sum(pmf[support <= 3])

p_interval = np.sum(pmf[(support >= 2) & (support <= 4)])

print(f"Normalizing constant c = {c:.4f}")
print(f"F(3) = {cdf_3:.4f}")
print(f"P(2 <= X <= 4) = {p_interval:.4f}")
```

```python
x_vals = np.array([1, 2, 3, 4, 5])
p_vals = x_vals / 15.0

expected_x = np.sum(x_vals * p_vals)
expected_x2 = np.sum((x_vals**2) * p_vals)
var_x = expected_x2 - (expected_x**2)

print(f"E[X]   = {expected_x:.4f}")
print(f"E[X^2] = {expected_x2:.4f}")
print(f"Var(X) = {var_x:.4f}")
```

```python
from scipy.stats import binom

n, p = 20, 0.25
# P(X >= 10) = 1 - P(X <= 9)
prob_pass = 1.0 - binom.cdf(9, n, p)

print(f"P(X >= 10) = {prob_pass:.6f}")
print(f"Mean = {binom.mean(n, p)}, Variance = {binom.var(n, p)}")
```

```python
from scipy.stats import geom, nbinom

p = 0.08

# 1. Geometric: P(X = 5)
p_geom_5 = geom.pmf(5, p)

# 2. Negative Binomial: scipy parameterizes by failures before r successes:
# 15 tests total with 3rd success on the 15th means 12 failures and r=3 successes.
p_nbinom = nbinom.pmf(12, 3, p)

print(f"P(1st defect on test 5): {p_geom_5:.5f}")
print(f"P(3rd defect on test 15): {p_nbinom:.5f}")
```

```python
from scipy.stats import hypergeom

# M = total population (20), n = total successes (12 red), N = sample size (6)
M, n, N = 20, 12, 6
prob_4_red = hypergeom.pmf(4, M, n, N)

print(f"P(X = 4) = {prob_4_red:.5f}")
```

```python
from scipy.stats import poisson

lam = 4.5
# P(3 <= X <= 6) = P(X <= 6) - P(X <= 2)
prob_interval = poisson.cdf(6, lam) - poisson.cdf(2, lam)

print(f"P(3 <= X <= 6) = {prob_interval:.5f}")
```

```python
import sympy as sp

x, k = sp.symbols('x k', real=True)
pdf_unscaled = 2*x - x**2

# Integrate over support [0, 2] to find normalizing constant
integral_val = sp.integrate(pdf_unscaled, (x, 0, 2))
k_val = 1 / integral_val
pdf = k_val * pdf_unscaled

# CDF at x = 1
cdf_1 = sp.integrate(pdf, (x, 0, 1))

print(f"k = {k_val}")
print(f"f(x) = {pdf}")
print(f"P(X <= 1) = {cdf_1.evalf():.4f}")
```

```python
mu = sp.integrate(x * pdf, (x, 0, 2))
ex2 = sp.integrate((x**2) * pdf, (x, 0, 2))
variance = ex2 - mu**2
std_dev = sp.sqrt(variance)

print(f"E[X]    = {mu.evalf():.4f}")
print(f"Var(X)  = {variance.evalf():.4f}")
print(f"Std Dev = {std_dev.evalf():.4f}")
```

```python
from scipy.stats import uniform

a, b = 0, 15
rv = uniform(loc=a, scale=b - a)

# P(X > 10)
p_gt_10 = 1.0 - rv.cdf(10)

# P(X > 12 | X > 8) = P(X > 12) / P(X > 8)
p_gt_12_given_gt_8 = (1.0 - rv.cdf(12)) / (1.0 - rv.cdf(8))

print(f"P(Wait > 10) = {p_gt_10:.4f}")
print(f"P(Wait > 12 | Wait > 8) = {p_gt_12_given_gt_8:.4f}")
```

```python
from scipy.stats import gamma

alpha = 4
beta = 2.0  # rate parameter (scipy uses scale = 1/beta)
rv = gamma(a=alpha, scale=1.0 / beta)

prob_le_3 = rv.cdf(3)
print(f"P(T_4 <= 3) = {prob_le_3:.5f}")
print(f"Gamma Mean: {rv.mean():.2f}, Variance: {rv.var():.2f}")
```

```python
from scipy.stats import norm

mu, sigma = 72, 9

# 1. P(60 <= X <= 85)
p_between = norm.cdf(85, mu, sigma) - norm.cdf(60, mu, sigma)

# 2. Percent-point function (inverse CDF)
p95_score = norm.ppf(0.95, mu, sigma)

print(f"P(60 <= Score <= 85) = {p_between * 100:.2f}%")
print(f"95th Percentile Cutoff = {p95_score:.2f}")
```

```python
import numpy as np

joint_pmf = np.array([
    [0.10, 0.20, 0.10],
    [0.15, 0.30, 0.15]
])

# Marginals
marginal_X = np.sum(joint_pmf, axis=1)  # shape (2,)
marginal_Y = np.sum(joint_pmf, axis=0)  # shape (3,)

# Independence check: P(X, Y) == P(X) * P(Y)
outer_product = np.outer(marginal_X, marginal_Y)
is_indep = np.allclose(joint_pmf, outer_product)

print(f"Marginal P(X): {marginal_X}")
print(f"Marginal P(Y): {marginal_Y}")
print(f"X and Y Independent? {is_indep}")
```
